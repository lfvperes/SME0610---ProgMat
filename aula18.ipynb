```python
import numpy as np
```

```python


# 1. Definindo a matriz A (coeficientes)
# Cada sub-lista representa uma linha da matriz
A = np.array([
    [0.2, 0.6, 1],
    [0.5, 0.4, 1],
    [0.4, 0.4, 1]
])

# 2. Definindo o vetor B (resultados)
B = np.array([0.56, 0.81, 0.46])

try:
    # 3. Resolvendo o sistema (Equivalente a x = A^-1 * B)
    # O NumPy usa decomposição LU, que é mais estável que inverter a matriz diretamente
    solucao = np.linalg.solve(A, B)

    x, y, z = solucao

    print("--- Solução do Sistema ---")
    print(f"x = {x:.2f}")
    print(f"y = {y:.2f}")
    print(f"z = {z:.2f}")

    # 4. Verificação (A * x deve ser igual a B)
    verificacao = np.allclose(np.dot(A, solucao), B)
    print(f"\nA solução é válida? {verificacao}")

except np.linalg.LinAlgError:
    # Caso as colunas não sejam Linearmente Independentes (L.I.), a matriz é singular
    print("Erro: A matriz é singular e não possui uma solução única.")
```

```python
l = [0, 0, 1]
a1 = [0.2, 0.6, 1]
a2 = [0.5, 0.4, 1]
a3 = [0.4, 0.4, 1]
c1 = 0.56 - np.dot(l, a1)
c2 = 0.81 - np.dot(l, a2)
c3 = 0.46 - np.dot(l, a3)
print(c1, c2, c3)
```
