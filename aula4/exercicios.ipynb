```python
from pulp import *
```

```python
problem1 = LpProblem("diversao_estudos", LpMaximize)
dec_vars = [LpVariable(f'x_{i}', lowBound=0, cat=LpContinuous) for i in range(1,3)]
a = [2, 1]
problem1 += lpSum(a[i] * dec_vars[i] for i in range(2))
b = [10, 4]
c = [[1, 1],
    [1, 0]]
for j in range(2):
    problem1 += lpSum([c[j][i] * dec_vars[i] for i in range(2)]) <= b[j]

status = problem1.solve(PULP_CBC_CMD(msg=False))
print([f'{dv}={dv.varValue}' for dv in dec_vars])
print(f'objective={sum([a[i] * dec_vars[i].varValue for i in range(2)])}')
```

```python
problem2 = LpProblem("dieta", LpMinimize)
dec_vars = [LpVariable(f'x_{i}', lowBound=0, cat=LpInteger) for i in range(1,3)]
a = [30, 25]
problem2 += lpSum(a[i] * dec_vars[i] for i in range(2))

problem2 += 30 * dec_vars[0] + 12 * dec_vars[1] >= 16
problem2 += 10 * dec_vars[0] + 1 * dec_vars[1] >= 6
problem2 += 1000 * dec_vars[0] + 2500 * dec_vars[1] <= 1800

status = problem2.solve(PULP_CBC_CMD(msg=False))
print([f'{dv}={dv.varValue}' for dv in dec_vars])
print(f'objective={sum([a[i] * dec_vars[i].varValue for i in range(2)])}')
```
