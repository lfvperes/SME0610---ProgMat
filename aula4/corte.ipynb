```python
from pulp import *
```

```python
def problema_corte(cat, a, b, c, n, m):
    problem = LpProblem("corte", LpMinimize)

    dec_vars = [LpVariable(f'x_{i}', lowBound=0, cat=cat) for i in range(1,n+1)]
    problem += lpSum([a[i] * dec_vars[i] for i in range(n)])

    for j in range(m):
        problem += lpSum([c[j][i] * dec_vars[i] for i in range(n)]) >= b[j]

    status = problem.solve(PULP_CBC_CMD(msg=False))

    for dv in dec_vars:
        print(f'{dv}={dv.varValue}')

    print(f'objective={sum([a[i] * dec_vars[i].varValue for i in range(n)])}')
```

```python
a = [1, 2, 3, 2]
c = [[8, 4, 2, 0],
    [0, 1, 2, 3]]
b = [500, 350]

n = len(a)
m = len(b)

problema_corte(LpInteger, a, b, c, n, m)
```
